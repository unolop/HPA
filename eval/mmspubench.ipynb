```python
import os 
import json 
from openai import OpenAI
from sklearn.model_selection import train_test_split 
from dotenv import load_dotenv

load_dotenv()
OPENAI_API_KEY = os.getenv('API_KEY')
client = OpenAI(api_key=OPENAI_API_KEY) # os.getenv("OPENAI_API_KEY"))
```

```python
system_prompt = """
You are a precise translation and data-formatting assistant.

CRITICAL FORMATTING RULES:
1. Translate the question into Korean
2. For options: Keep the EXACT format "A: <English> <Korean>, B: <English> <Korean>, ..."
3. Separate options with a comma and a single space
4. Separate English and Korean with a single space
5. DO NOT use comma within an option but only use one between the options
6. Numbers should appear in both languages if they appear in the text

Required JSON structure:
{
  "question_kr": "<Korean translation of question>",
  "options": "<Formatted options string>"
}

Example of CORRECT options format:
"A: In her hand 그녀의 손에, B: On her shoulder 그녀의 어깨에, C: On the ground 땅에, D: Inside the man's bag 그 남자의 가방 안에"

Return ONLY the JSON dictionary with no additional text or markdown formatting.
"""

def preprocess(question, options): 
  user_prompt = f"""
  QUESTION: {question}
  OPTIONS_STRING: {options}
  """

  response = client.chat.completions.create(
      model="gpt-5-nano",
      messages=[
          {"role": "system", "content": system_prompt},
          {"role": "user", "content": user_prompt}
      ],
      temperature=1,  # Changed to 0 for consistent formatting
  )
  output = response.choices[0].message.content.strip()
  parsed = json.loads(output)
  print(user_prompt, parsed)
  return parsed['question_kr'], parsed['options']
```
