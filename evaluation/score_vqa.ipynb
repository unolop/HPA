```python
def read_file(input_path): 
    
    data = []
    with open(input_path, 'r', encoding='utf-8') as f:
        for line in f:
            if line.strip():
                data.append(json.loads(line)) 

    return data 
    
data = read_file(files_textvqa[1])
```
