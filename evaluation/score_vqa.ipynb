```python
from analysis.utils.score import get_summary, extract_mc_choice, MODEL_DISPLAY_MAP 

okvqa = get_summary('okvqa').drop_duplicates(subset=['condition', 'question_id', 'model']) 
okvqa.pivot_table(index='model', columns='condition', values=['question_id', 'correct'], aggfunc='nunique')
okvqa.pivot_table(index='model', columns='condition', values=['correct'], aggfunc='mean')   
```

```python
correct_qids = okvqa[(okvqa['correct'] == 100) & (okvqa['question_type'] == okvqa.question_type.unique()[1])].question_id.unique() 
df = (okvqa[okvqa['question_id'] == correct_qids[1]]
      .sort_values(['correct', 'condition']))
print(df.iloc[0]['question'], '\n',
      'Answers:',df.iloc[0]['answers'], '\n',
      'answer_type: ', df.iloc[0]['answer_type'],  '\n',
      'question_type: ', df.iloc[0]['question_type'],  )
pivoted = df.pivot(index=['condition'], columns='model', values='output') 
pivoted  
```

```python
okvqa_summary = okvqa.pivot_table(index=['answer_type', 'question_type'], columns=['condition'], values=['correct'], aggfunc='mean')
okvqa_summary = okvqa_summary.sort_values([('correct', 'inst blind')], ascending=False)
okvqa_summary 
```
