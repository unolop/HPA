```python
def swift_dataset(vqa_data, models='llm', blind=True, system_message=system_message):
    output_path = f"/home/work/yuna/HPA/data/swift/{models}_vqav2_val_1k"
    if system_message is not None:
        output_path += "_inst"
    
    if blind:
        output_path += "_blind"
    output_path += ".jsonl"
    
    with open(output_path, 'w', encoding='utf-8') as f:
        for item in vqa_data:
            question = system_message  + item['question']
            if not models == 'llm':
                question = f"<image>\n{question}"
            messages = []

            # Only add system message if it's not None
            # if system_message is not None:
            #     messages.append({
            #         "role": "system",
            #         "content": "system_message"
            #     })
            
            messages.append({
                "role": "user",
                "content": question
            })
            
            swift_item = {
                "id": str(item.get('question_id', item.get('image_id', ''))),
                "messages": messages
            }
            
            if not models == 'llm':
                if blind:
                    swift_item['image'] = ["/home/work/yuna/HPA/data/blank_224.png"]
                else:
                    image_path = os.path.join(image_folder, item['image_id'])
                    swift_item['image'] = [image_path]
            
            f.write(json.dumps(swift_item, ensure_ascii=False) + '\n')
    
    print(f"saved to {output_path}")
```
